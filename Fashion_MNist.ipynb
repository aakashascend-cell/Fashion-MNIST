```python
# Download latest version
path = kagglehub.dataset_download("zalando-research/fashionmnist")
```

```python
(x_train,y_train),(x_test,y_test) = keras.datasets.fashion_mnist.load_data()
```

```python
x_train = x_train.reshape(60000,784)
x_test =x_test.reshape(10000,784)
x_train = x_train.astype('float32')
x_test = x_test.astype('float32')
x_train /= 255
x_test /= 255
```

```python
y_train = keras.utils.to_categorical(y_train,10)
y_test = keras.utils.to_categorical(y_test,10)
```

```python
batch = 128
epoch = [20,30,40]
```

```python
model = Sequential()
model.add(Dense(512, input_shape=(784,), activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(512, activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(10, activation='softmax'))
```

```python
model.compile(optimizer='adam', metrics = ['accuracy'], loss = 'categorical_crossentropy')
```

```python
model_files = {
    20: "model_epoch_20.keras",
    30: "model_epoch_30.keras",
    40: "model_epoch_40.keras"
}
results = {}
```

```python
# Dictionary to log accuracy metrics
accuracy_log = {}


for target_epoch in epoch:
    print(f"\n================ Training for Epoch {target_epoch} ================")

    # 1. Fit the model to the next target epoch
    history = model.fit(
        x_train,
        y_train,
        batch_size=batch,
        epochs=target_epoch,

        verbose=1,
        validation_data=(x_test, y_test)
    )

    # 2. Save the trained model file at this milestone
    model_filename = f"model_epoch_{target_epoch}.keras"
    model.save(model_filename)
    print(f"--> Saved model to {model_filename}")


    current_train_acc = history.history['accuracy'][-1]
    current_val_acc = history.history['val_accuracy'][-1]

    accuracy_log[target_epoch] = {
        'train_acc': current_train_acc,
        'val_acc': current_val_acc
    }

    print(f"--> Milestone Epoch {target_epoch} Metrics:")
    print(f"    Train Accuracy: {current_train_acc:.4f}")
    print(f"    Val Accuracy:   {current_val_acc:.4f}")



# ================= Final Summary =================
print("\nSummary of Saved Milestones:")
for ep, accs in accuracy_log.items():
    print(f"Epoch {ep} -> Train Acc: {accs['train_acc']:.4f} | Val Acc: {accs['val_acc']:.4f}")
```
